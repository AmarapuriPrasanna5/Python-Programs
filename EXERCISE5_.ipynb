```python
#function that return multiple values
def mul_fun(*num):
  s=sum(num)
  c=len(num)
  avg=s/c
  return s,c,avg
s,count,average = mul_fun(1,5,7,3,9,6)
print(s,count,average )
```

```python
#Program that reverse a string using user defined functions
def reverse(x):
   #for i in range(len(x)-1,-1,-1): x = x+s[i] return x
    rev = x[::-1]
    return rev
x = input()
reverse(x)
```
