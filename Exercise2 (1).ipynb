```python
# Program that finds maximum among 3 numbers
a = int(input("Enter number "))
b = int(input("Enter  another number "))
c = int(input("Enter another number "))
if a>b and a>c:
  print(f"{a} is maximum ")
elif a<b and b>c:
  print(f"{b} is maximum ")
else :
  print(f"{c} is maximum ")
```

```python
# given year is leap year or not
year = int (input("Enter a year "))
if year%4==0:
  if year%100!=0 or year%400==0:
    print(" leap year ")
else :
  print("Not leap year ")
```

```python
#Example code on Match Case
day =int(input())
match day:
  case 1:
    print("Sunday")
  case 2:
    print("monday")
  case 3:
    print("Tuesday")
  case 4:
    print("Wednesday")
  case 5:
    print("thrusday")
  case 6:
    print("Friday")
  case 7:
    print("Saturday")
  case _:
    print("Invalid")
```

```python
#Sum of the Digits of a number
n = int(input("Enter the number to sum: "))
sum=0
while(n!=0):
  rem = n%10
  sum+=rem
  n = n//10
print(sum)
```

```python
# Reverse of a number
n = int(input("Enter the number to sum: "))
sum=0
while(n!=0):
  rem = n%10
  sum=sum*10 + rem
  n = n//10
print(sum)
```

```python
#Prime or not
n = int(input("Enter the number: "))
for i in range(2,n):
  if n%i==0:
    print("Not Prime")

    break
else :
    print("Prime")
```
