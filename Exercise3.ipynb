```python
#list operations
# concatenate
l1=[1,2,3,4,"jay"]
l2=[7,4,9,0,2,8]
print(l1+l2)
```

```python
#Repetition
print(l2*3)
```

```python
#Indexing
print(l1[3])
print(l2[-4])
```

```python
#Slicing
print(l1[0:4])
print(l2[::-2])
```

```python
#membershp operator
if 22  not in l2:
  print("yeah ")
else:
  print("no")
```

```python
#Iteration
#for i in l2:
    #print(i)
for i in range(len(l2)):
  print(l2[i])
```

```python
#Built-in functions
l3=[45,7,2,90,67,23,78,453,982]
l4 = [1,2,3,4,5,6,7,8,9]
print(sum(l3))
print(max(l3))
print(min(l4))
print(len(l4))
print(sorted(l4))
print(list(reversed(l3)))
```

```python
#list methods (built in methods)
#append
l2.append(5)
print(l2)
```

```python
#extend
l6=[3,5,67,89,23,68,99]
l7=[7,3,9,77,88,889]
l6.extend(l7)
print(l6)
```

```python
#insert
l7.insert(0,"hlo")
print(l7)
```

```python
#count
print(l3.count(7))
```
