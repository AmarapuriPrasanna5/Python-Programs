```python
#creating dictionaries
s = {'sno':1,'name':'siri'}
m = dict(x=5,y=7)
print(m)
```

```python
#accesing values
print(s['sno'])
s['age']=20
s['name']='sai'
print(s)
```

```python
#Built-in functions
print(len(s))
print(sorted(s))#sorted in key order not value
print(max(s))
print(min(s))
print(str(s))
```

```python
#methods in dictionaries
print(s.get('name'))
print(s.keys())
print(s.values())
print(s.items())
print(s.update({"Branch":"CSE"}))
print(s.pop('name'))
print(s.setdefault('age',20))
```

```python
#A program to check if the given key is in dictionary or not.
x = {'Name':'Kiran','Age':23,'branch':'AI&DS' }
key=input()
if key in x.keys():   # or directly x
  print('True')
else:
  print("False")
```
